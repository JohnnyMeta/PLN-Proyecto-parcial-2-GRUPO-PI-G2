#3) Plantillas de las Funciones

```python
def crear_bloque_conocimiento(id,
                              nombre,
                              materias):

    """Bloque de conocimiento que contiene información de materias evaluadas
    (% y número de preguntas) en el examen de admisión."""

    return {
        "id": id,
        "nombre": nombre,
        "materias": materias
        }


def crear_ficha_carrera(id,
                        facultad_id,
                        bloque_id,
                        titulo_otorgado=None,
                        duracion=None,
                        modalidad=None,
                        sede=None,
                        descripcion=None,
                        perfil_empleabilidad=None):

    """
    Ficha informativa completa de una carrera,
    None, existe para reutilización del formato con información independiente.
    """

    return {
        "id": id,
        "facultad_id": facultad_id,
        "bloque_id": bloque_id,
        "titulo_otorgado": titulo_otorgado,
        "duracion": duracion,
        "modalidad": modalidad,
        "sede": sede,
        "descripcion": descripcion,
        "perfil_empleabilidad": perfil_empleabilidad,
        "ficha_completa": all([titulo_otorgado, duracion, modalidad, sede, descripcion, perfil_empleabilidad]),
    }


def crear_fecha_clave(id,
                      nombre,
                      fecha_inicio,
                      fecha_fin,
                      descripcion):

    """Periodo oficial del cronograma de admisión (formato AAAA-MM-DD)."""

    return {
        "id": id,
        "nombre": nombre,
        "fecha_inicio": fecha_inicio,
        "fecha_fin": fecha_fin,
        "descripcion": descripcion
        }
```
